#### NumPy 배열 예제
실습에서 사용한 가상의 성적 데이터를 저장한 배열을 다시 살펴보자 

```python
import numpy as np
import pandas as pd

subjects = ["과목A", "과목B", "과목C", "과목D"]

scores = np.array([[85, 72, 90, 68],
                   [92, 88, 95, 91],
                   [45, 58, 62, 40],
                   [78, 95, 71, 83],
                   [88, 81, 84, 90],
                   [60, 47, 55, 72],
                   [95, 69, 78, 66],
                   [70, 90, 88, 59]])

print(scores)
```

- 3번 학생의 과목C 점수를 출력하고 싶다면 어떻게 해야할까?

























```python
print("3번 학생의 과목C 점수:", scores[3, 2])
```

- NumPy 배열에서는 행과 열을 **번호로만** 지칭할 수 있음
- `scores[3, 2]` 가 무엇을 뜻하는지 알려면 데이터 설명을 따로 찾아봐야함
- 열의 순서가 바뀌면 코드가 틀린 값을 내놓게 됨

같은 데이터를 pandas의 DataFrame으로 만들어 보자

```python
df = pd.DataFrame(scores,
                  columns=["과목A", "과목B", "과목C", "과목D"],
                  index=["학생" + str(i) for i in range(8)])
df
```

```python
# 번호가 아니라 이름으로 지칭할 수 있다
print(df["과목C"])
print()
print("과목별 평균:")
print(df.mean())
```

- 값은 똑같지만 **행과 열에 이름표가 붙음**
- `scores[:, 2]` 대신 `df["과목C"]` 라고 쓸 수 있음

- Series 객체의 덧셈 연산

```python
import pandas as pd

# A, B, C, D 네 회사의 1차년도 매출액과 2차년도 매출액이 주어졌다고 가정

sales = {'A': 5000000, 'B': 7100000, 'C': 3600000, 'D': 4050000}
year1 = pd.Series(sales)

year2 = pd.Series([3500000, 5500000, 4000000, 6000000], index=['B','D','C','A'])

print("year1 :")
print(year1)
print()
print("year2 :")
print(year2)
```

#### [예측] 실행하기 전에 생각해 볼 것

- `year1`은 **A, B, C, D 순서**이고 `year2`는 **B, D, C, A 순서**임
- 두 Series를 그냥 `year1 + year2` 로 더하면 어떤 결과가 나올까?

  1. 순서대로 위에서부터 더해진다 (A+B, B+D, C+C, D+A)
  2. 같은 이름끼리 찾아서 더해진다 (A+A, B+B, C+C, D+D)
  3. 순서가 달라서 오류가 난다

- NumPy 배열이었다면 어떻게 되었을지도 함께 생각해 볼 것



```python
total = year1 + year2
print(total)
```

- 기본적인 통계 함수
  - sum(), mean(), median(), var(), std(), count(), max(), min(), argmax(), argmin()

```python
print('네 회사의 2년간 매출액 합계: ', total.sum())
print('네 회사의 2년간 매출액 평균: ', total.mean())
print('네 회사의 2년간 매출액 최대값: ', total.max())
print('네 회사의 2년간 매출액 최소값: ', total.min())
print('2년간 매출액 최대값의 인덱스 번호: ', total.argmax())
print('2년간 매출액이 가장 많은 회사: ', total.index[total.argmax()])
print('2년간 매출액 최소값의 인덱스 번호: ', total.argmin())
print('2년간 매출액이 가장 적은 회사: ', total.index[total.argmin()])

# idxmax(), idxmin() 는 위치 대신 인덱스 이름을 바로 돌려줌
print()
print('idxmax()로 구한 매출액 1위 회사: ', total.idxmax())
print('idxmin()로 구한 매출액 최하위 회사: ', total.idxmin())
```

#### DataFrame
- 표와 같은 자료 구조
- 여러 개의 행(row)과 열(column)로 구성됨
  - 각 열은 서로 다른 종류의 값(숫자, 문자열, 불리언 등)을 담을 수 있음
- 행과 열에 대한 인덱스를 가지고 있음

- DataFrame 객체 생성
  - dictinary의 key, value 중 value에 리스트를 사용

```python
import pandas as pd

data = {'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada', 'Nevada'],
        'year': [2000, 2001, 2002, 2001, 2002, 2003],
        'pop': [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}

frame = pd.DataFrame(data)
frame
```

- 원하는 열의 순서로 DataFrame 객체 생성 가능

```python
frame2 = pd.DataFrame(data, columns=['year', 'state', 'pop', 'debt'],
                      index=['one', 'two', 'three', 'four', 'five', 'six']) 

frame2
```

- 열에 대한 인덱스로 해당 열에 대한 데이터만 추출

### 파생 변수 만들기
- 데이터에 들어 있는 변수만 이용해 분석할 수도 있지만, 변수를 조합하거나 새 변수를 만들어 분석하는 것도 가능
  - 예를 들어 여러 과목의 시험 점수를 조합해 전과목 평균 점수를 나타내는 변수를 만들어 분석할 수 있음
  - 기존의 변수를 변형해 만든 변수를 파생 변수라고 함

```python
df = pd.DataFrame({'var1': [4,3,8],
                  'var2': [2,6,1]})
df
```

- var1과 var2 변수의 값을 더한 var_sum이라는 파생 변수를 만들어 df에 추가하기

```python
df['var_sum'] = df['var1'] + df['var2']

df
```

- var1과 var2 변수의 평균을 계산하여 var_mean이라는 파생 변수를 만들어 df에 추가하기

```python
df['var_mean'] = (df['var1'] + df['var2']) / 2

df
```

- 위 방식은 열이 두세 개일 때는 편하지만, **열이 많아지면 일일이 나열하기 어려움**
  - 예를 들어 과목이 7개라면 `df['A'] + df['B'] + ...` 로 일곱 번을 적어야 함
- 이때는 지난 시간 NumPy에서 배운 **`axis`** 를 그대로 사용 가능
  - `axis=1` : 열 방향으로 계산 -> **행(관측치)마다** 결과가 나옴
  - `axis=0` : 행 방향으로 계산 -> **열(변수)마다** 결과가 나옴
  - `axis` 를 생략하면 기본값은 `axis=0`

```python
# 앞에서 만든 파생 변수와 같은 결과를 axis 로 계산
df['var_sum2']  = df[['var1', 'var2']].sum(axis=1)
df['var_mean2'] = df[['var1', 'var2']].mean(axis=1)

df
```

- `var_sum` 과 `var_sum2`, `var_mean` 과 `var_mean2` 의 값이 동일함
- 열이 많아질수록 아래 방식이 훨씬 간결함

- 계산 대상 열을 `df[['var1', 'var2']]` 처럼 **명시적으로 골라 주는 것이 중요**
  - `df.sum(axis=1)` 로 전체를 쓰면 앞에서 만든 파생 변수(`var_sum` 등)까지 함께 더해짐
  - 오류는 나지 않고 값만 틀리게 나오므로 발견하기 어려움
- 아래 셀에서 그 차이를 확인해 볼 것

```python
print("계산할 열만 지정:")
print(df[['var1', 'var2']].sum(axis=1))
print()
print("전체를 대상으로 계산 (파생 변수까지 포함되어 값이 달라짐):")
print(df.sum(axis=1))
```

- `axis=0` 으로 계산하면 **변수(열)별** 결과가 나옴

```python
print("변수별 평균 (axis=0):")
print(df[['var1', 'var2']].mean(axis=0))
```
