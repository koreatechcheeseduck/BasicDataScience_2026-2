### 데이터 시각화 (Data visualization)
- 데이터를 그래프나 차트 등 시각적인 형태로 표현하여 데이터를 쉽게 이해할 수 있도록 하는 것
  - 데이터 원자료나 통계표는 수많은 숫자와 문자로 구성되어 있어서 내용을 한눈에 파악하기 어려움
- 데이터를 그래프로 표현하면 데이터의 추세와 경향성, 상관관계 등이 잘 드러나기 때문에 데이터의 특징을 쉽게 이해할 수 있음
- 효과적인 시각화는 데이터 분석 결과를 직관적으로 전달하고 의사결정에 필요한 인사이트를 제공함
- 주요 도구로는 matplotlib, seaborn, Plotly, Tableau 등이 있음

### 시각화 그래프 종류의 선택
- 그래프는 종류는 **데이터에서 무엇을 확인하고 싶은지**에 따라 고름

| 보고 싶은 것 | 질문의 예 | 주로 쓰는 그래프 | 
|---|---|---|
| 추세 | 시간에 따라 어떻게 변하는가? | 선 그래프 | 
| 분포 | 값들이 어떻게 퍼져 있는가? | 히스토그램, 밀도 그래프 | 
| 그룹 간 비교 | 그룹마다 분포나 평균이 다른가? | 상자 그림, 바이올린, 막대 그래프 | 
| 관계 | 두 변수가 함께 변하는가? | 산점도, 회귀선 | 
| 여러 변수 | 세 개 이상의 변수를 함께 보려면? | 색상 구분, 나눠 그리기, heatmap | 

- 이번 주
  - matplotlib으로 **그래프의 기본 구조**를 익힘
  - seaborn으로 **분포와 그룹 간 비교**를 다룸

### matplotlib 라이브러리

#### 데이터 시각화의 가장 기본이 되는 파이썬 라이브러리
- 주로 2D 그래프를 생성하기 위한 라이브러리
  - 출판물 수준의 그래프를 만들어내도록 설계됨
  - https://matplotlib.org/
- PDF, SVG, JPG, PNG 등 일반적으로 널리 사용되는 형식으로 그래프를 저장할 수 있음
- 뒤에서 다룰 seaborn도 matplotlib 을 기반으로 만들어졌음
  - 그래서 seaborn으로 그린 그래프의 제목, 축 이름 등은 matplotlib 함수로 설정함

#### matplotlib의 pyplot 모듈
- 데이터 시각화를 위한 명령어 형태의 함수 제공
- 보통 `plt` 라는 이름으로 불러와서 사용

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]       # x축에 사용될 값
y = [20, 30, 40]    # y축에 사용될 값

plt.plot(x, y)      # plot() - 선 그래프를 그림
plt.show()          # 그린 그래프를 화면에 출력
````

#### 선 그래프 (Line graph)
- 시간에 따른 변화나 연속적인 데이터를 시각화하는 데 적합함
- 연속적인 추세나 패턴을 파악하는 데 용이함

##### 선 그래프로 시각화하는 데 적합한 데이터 예
- 시간 혹은 날짜에 따른 날씨(온도, 강수량)
- 월별 제품 판매량
- 일별 주식 가격
- 연도별 인구수

##### 그래프 꾸미기
- `title( )`, `xlabel( )`, `ylabel( )`: 제목과 축 이름
- `label` 매개변수와 `legend( )`: 범례
- `linestyle`, `marker` 매개변수: 선 모양과 점 모양
- `grid( )`: 격자선

````python
import matplotlib.pyplot as plt

# 가상의 계절별 시간대 기온 데이터
hour   = [6, 9, 12, 15, 18, 21]
summer = [15, 18, 22, 25, 21, 18]
winter = [1, 3, 6, 7, 4, 2]

plt.plot(hour, summer, label='Summer', marker='o')
plt.plot(hour, winter, label='Winter', marker='s', linestyle='--')

plt.title('Temperature by Hour', fontsize=14)
plt.xlabel('Hour') # x축 이름
plt.ylabel('Temperature (C)') # y축 이름
plt.legend()          # plot() 함수에서 label 매개변수로 지정한 이름으로 범례 표시
plt.grid(True)        # 격자선 표시

plt.show()
````

- `plot( )`을 여러 번 호출하면 **같은 그래프 위에 겹쳐서** 그려짐
- `plt.show( )`를 만나면 지금까지 그린 내용이 출력되고, 그 뒤부터는 새 그래프가 시작됨

#### 여러 그래프를 나란히 그리기
- 그래프를 비교할 때는 나란히 놓고 보는 것이 편함
- `plt.subplots(행 수, 열 수)`는 그래프를 그릴 칸을 여러 개 만들어 줌
  - 첫 번째 반환값 `fig`: 전체 그림
  - 두 번째 반환값 `axes`: 각 칸. 칸마다 따로 그래프를 그릴 수 있음
- 칸에 그릴 때는 `plt.title( )` 대신 `칸.set_title( )`처럼 `set_` 이 붙은 함수를 사용

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))   # 1행 2열, 전체 크기 가로 10 세로 4

axes[0].plot(hour, summer, marker='o')
axes[0].set_title('Summer')

axes[1].plot(hour, winter, marker='s', color='tab:orange')
axes[1].set_title('Winter')

plt.show()
````

- 칸마다 y축 범위가 자동으로 정해지므로, 값의 수준이 달라도 그래프 모양만 보면 비슷해 보일 수 있음
- 나란히 놓고 **값을 비교할 때는** subplots() 함수에서 `sharey=True`로 y축 범위를 같게 맞출 것

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True)   # 1행 2열, 전체 크기 가로 10 세로 4

axes[0].plot(hour, summer, marker='o')
axes[0].set_title('Summer')

axes[1].plot(hour, winter, marker='s', color='tab:orange')
axes[1].set_title('Winter')

plt.show()
````

#### 그래프를 이미지 파일로 저장
- `savefig( )` 함수
  - 저장 가능 형식: png, pdf, svg 등
  - `dpi`: 해상도. 보고서용은 보통 300
- **`plt.show( )`보다 먼저** 호출해야 함. `show( )` 뒤에 저장하면 빈 그림이 저장됨

````python
plt.plot(hour, summer, marker='o')
plt.title('Summer Temperature')

plt.savefig('summer.png', dpi=300)   # 현재 폴더에 이미지 파일로 저장
plt.show()
````

#### 그래프에 한글 사용하기
- matplotlib 의 기본 글꼴은 한글을 지원하지 않아 한글이 네모로 표시됨
- 운영체제마다 설치된 한글 글꼴이 다르므로 환경에 맞게 설정해야 함

| 환경 | 글꼴 이름 |
|---|---|
| Windows | `'Malgun Gothic'` (맑은 고딕) |
| Mac | `'AppleGothic'` |
| Colab | 기본으로 설치된 한글 글꼴이 없어 설치가 필요함 (아래 참고) |

- `axes.unicode_minus`를 `False`로 설정하지 않으면 음수 부호가 깨짐
- 이 자료의 나머지 예제는 환경에 관계없이 실행되도록 **그래프 글자를 영문으로** 작성함

````python
import platform
import matplotlib.pyplot as plt

system = platform.system()
if system == 'Windows':
    plt.rc('font', family='Malgun Gothic')
elif system == 'Darwin':          # Mac
    plt.rc('font', family='AppleGothic')
else:
    print("Windows, Mac 이 아닌 환경임. Colab 이라면 아래 설명을 참고하여 글꼴을 설치할 것")

plt.rcParams['axes.unicode_minus'] = False   # 음수 부호 깨짐 방지

plt.plot(hour, summer, marker='o')
plt.title('시간대별 기온')
plt.xlabel('시각')
plt.ylabel('기온')
plt.show()
````

### seaborn 라이브러리

#### matplotlib을 기반으로 한 통계 그래프 전용 라이브러리
- 짧은 코드로도 복잡한 그래프를 그릴 수 있음
- pandas의 DataFrame을 그대로 사용할 수 있음
  - `data=데이터프레임, x='열 이름', y='열 이름'` 형태로 지정
  - 축 이름도 열 이름으로 자동 표시됨
- https://seaborn.pydata.org/

#### 실습 데이터: 펭귄 데이터
- 6주차 종합 실습에서 사용한 데이터
- seaborn 에 내장된 데이터는 `sns.load_dataset( )`으로 불러올 수 있음 (인터넷 연결 필요)

| 변수 | 의미 |
|---|---|
| `species` | 종 (Adelie, Chinstrap, Gentoo) |
| `island` | 관측된 섬 (Biscoe, Dream, Torgersen) |
| `bill_length_mm` | 부리 길이 (mm) |
| `bill_depth_mm` | 부리 깊이 (mm) |
| `flipper_length_mm` | 지느러미 길이 (mm) |
| `body_mass_g` | 체중 (g) |
| `sex` | 성별 |

````python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins')

print(penguins.shape)
print(penguins.head())
````

### 분포 보기 - 한 변수의 값이 어떻게 퍼져 있는가

#### 히스토그램 (Histogram)
- 연속형 데이터의 값을 구간(bins)으로 나누어 각 구간에 값이 몇 개 있는지 막대로 표현
- 데이터의 분포 형태, 중심, 퍼진 정도, 이상치를 파악하는 데 유용함

##### 히스토그램을 활용하는 경우
- 연속형 데이터의 값을 여러 구간으로 나누어 빈도수를 확인하는 경우
  - 시험 점수, 나이, 키, 몸무게, 온도
- 데이터가 특정 구간에 몰려 있는지, 고르게 퍼져 있는지 확인하는 경우
- 이상치를 찾는 경우

##### matplotlib 과 seaborn 비교
- 같은 히스토그램을 두 라이브러리로 그려 봄

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# matplotlib - 결측치를 직접 제거해야 하고, 축 이름도 직접 지정해야 함
axes[0].hist(penguins['body_mass_g'].dropna(), bins=20)
axes[0].set_title('matplotlib')
axes[0].set_xlabel('body_mass_g')
axes[0].set_ylabel('Count')

# seaborn - DataFrame과 열 이름만 지정하면 됨
sns.histplot(data=penguins, x='body_mass_g', bins=20, ax=axes[1])
axes[1].set_title('seaborn')

plt.show()
````

- 두 그래프의 모양은 같음
- seaborn 은 결측치를 자동으로 제외하고, 축 이름도 열 이름으로 표시해 줌
- seaborn 함수에 `ax=칸`을 지정하면 원하는 칸에 그릴 수 있음
- **이후로는 seaborn 을 중심으로** 사용하고, 제목과 축 이름은 matplotlib 함수로 설정함

##### 구간(bins) 수에 따라 달라지는 모양
- 히스토그램은 구간을 어떻게 나누느냐에 따라 모양이 크게 달라짐
  - 구간이 너무 적으면 세부 모양이 뭉개짐
  - 구간이 너무 많으면 들쭉날쭉해져 전체 모양을 보기 어려움

````python
fig, axes = plt.subplots(1, 3, figsize=(14, 4)) # 1개 행 3개 열. 가로로 3개의 그래프를 넣을 수 있음

for ax, b in zip(axes, [5, 20, 80]):
    sns.histplot(data=penguins, x='body_mass_g', bins=b, ax=ax)
    ax.set_title(f'bins = {b}')

plt.show()
````

#### 밀도 그래프 (KDE plot)
- 히스토그램의 막대를 부드러운 곡선으로 이은 형태
  - 커널 밀도 추정(kernel density estimation) 그래프라고도 함
- 구간을 정하지 않아도 되므로, 여러 분포를 겹쳐서 비교할 때 편리함
- `histplot( )`에 `kde=True`를 지정하면 히스토그램 위에 함께 그릴 수 있음

````python
sns.histplot(data=penguins, x='body_mass_g', kde=True)

plt.title('Distribution of Body Mass')
plt.show()
````

### 그룹 간 비교 - 그룹마다 분포가 다른가

#### 그룹별 분포 겹쳐 그리기
- `hue` 매개변수에 범주형 변수를 지정하면 그룹마다 다른 색으로 그림

````python
sns.kdeplot(data=penguins, x='body_mass_g', hue='species', fill=True) # fill 매개변수 값을 True로 한 경우와 False로 한 경우 비교

plt.title('Body Mass by Species')
plt.show()
````

- Adelie 와 Chinstrap 은 거의 겹치고, Gentoo 만 오른쪽으로 떨어져 있음
- 전체 히스토그램만 보았을 때는 알기 어려웠던 정보

#### 상자 그림 (Box plot)
- 데이터의 다섯 가지 요약 값(최솟값 쪽 수염 끝, 1사분위수, 중앙값, 3사분위수, 최댓값 쪽 수염 끝)과 극단치를 한눈에 보여줌
- **여러 그룹의 분포를 나란히 비교**할 때 특히 유용함
  - 남성과 여성의 키 분포 비교, 여러 회사의 연봉 분포 비교 등
- 상자와 수염을 읽는 방법은 지난 시간 데이터 정제에서 다룬 내용과 같음

````python
sns.boxplot(data=penguins, x='species', y='body_mass_g', hue='species') # hue 매개변수 값을 지정한 경우와 그렇지 않은 경우 비교

plt.title('Body Mass by Species')
plt.show()
````

#### 바이올린 그래프 (Violin plot)
- 상자 그림에 밀도 그래프를 함께 표현한 그래프
- 상자 그림은 요약 값만 보여주지만, 바이올린 그래프는 **분포의 모양**까지 보여줌

### 아래 셀을 실행하기 전에 생각해 볼 것

- 6주차 종합 실습에서 **Gentoo 가 다른 두 종보다 몸집이 훨씬 크다**는 것을 확인했음
  - 지느러미 길이도 마찬가지로 Gentoo가 김
- 세 종을 구분하지 않고 **모두 섞어서** 지느러미 길이의 분포를 그리면 어떤 모양이 될까?
  1. 가운데가 볼록한 봉우리 하나
  2. 봉우리 두 개
  3. 전체적으로 평평한 모양
- 그리고 그 모양이 **상자 그림에서도 보일지** 생각해 볼 것

> 나의 예측:

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

sns.boxplot(data=penguins, y='flipper_length_mm', ax=axes[0])
axes[0].set_title('Box plot')

sns.violinplot(data=penguins, y='flipper_length_mm', ax=axes[1])
axes[1].set_title('Violin plot')

plt.show()
````

### 그룹 간 비교 - 개수와 평균

#### 막대 그래프 (Bar graph)
- 범주형 데이터나 여러 항목의 값을 비교할 때 적합함
- 막대의 길이로 항목 간의 상대적인 차이를 쉽게 파악할 수 있음

##### 막대 그래프로 시각화하는 데 적합한 데이터 예
- 국가별 인구
- 제품 종류별 판매량
- 브랜드별 시장 점유율
- 설문 조사에서 응답자의 선택 비율

##### seaborn 의 두 가지 막대 그래프

| 함수 | 막대 높이 | 비교 |
|---|---|---|
| `countplot( )` | 각 범주의 **개수** | `value_counts( )` 와 같은 계산 |
| `barplot( )` | 각 범주의 **평균** (기본값) | `groupby( ).agg( )` 와 같은 계산 |

````python
sns.countplot(data=penguins, x='island', hue='species')

plt.title('Number of Penguins by Island')
plt.show()
````

- 6주차 종합 실습에서 표로 확인했던 내용을 그래프로 보면 한눈에 드러남
  - Adelie는 세 섬 모두에 있음
  - Gentoo는 Biscoe에만, Chinstrap 은 Dream에만 있음

````python
sns.barplot(data=penguins, x='species', y='body_mass_g')

plt.title('Mean Body Mass by Species')
plt.show()

# 같은 값을 groupby 로 확인
print(penguins.groupby('species').agg(mean_mass=('body_mass_g', 'mean')).round(1))
````

- `barplot( )`은 그래프를 그리면서 **평균을 직접 계산**함
  - 위 표의 값과 막대 높이가 같은지 확인할 것
- 막대 위의 **검은 세로선**은 평균의 신뢰구간(오차 막대)
  - 평균이 대략 이 범위 안에 있을 것이라는 불확실성의 정도를 나타냄
  - 개수가 적거나 값이 많이 흩어진 그룹일수록 선이 길어짐
  - 필요 없으면 `errorbar=None` 으로 지울 수 있음
- 평균 대신 다른 값을 쓰려면 `estimator` 매개변수 사용 (예: `estimator='median'`)

##### 참고 - 파이 그래프
- 전체에서 각 항목이 차지하는 비율을 원의 조각으로 표현하는 그래프 (`plt.pie( )`)
- 사람의 눈은 **각도나 넓이보다 길이를 더 정확하게 비교**하므로, 비율을 비교할 때도 막대 그래프가 더 정확함
- 항목이 2~3개뿐이고 "전체 대비 일부"를 강조할 때에만 제한적으로 사용할 것
